```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Data from the user's request
data = {
    'Algorithm': ['RandomForest', 'LightGBM', 'XGBoost', 'SVM', 'LogisticRegression'],
    'Accuracy': [65.17, 67.87, 67.51, 56.43, 52.03]
}
df = pd.DataFrame(data)

# Sort the data by Accuracy in descending order
df = df.sort_values(by='Accuracy', ascending=False).reset_index(drop=True)

# Set the style
sns.set_style("whitegrid")

# Define custom colors
custom_palette = ['orange', 'purple', 'lightgreen', 'yellow', 'blue']


# Create the figure and the axes
plt.figure(figsize=(10, 6))
ax = sns.barplot(x='Algorithm', y='Accuracy', data=df, palette=custom_palette)

# Add title and labels
plt.title('Accuracy Comparison', fontsize=14)
plt.ylabel('Accuracy (%)', fontsize=12)
plt.xlabel('Algorithm', fontsize=12)

# Set y-axis limits to better show the differences
plt.ylim(50, 70)
ax.yaxis.grid(True, linestyle='--', alpha=0.7)
ax.xaxis.grid(False)

# Add the accuracy values on top of the bars
for i, row in df.iterrows():
    accuracy_value = f"{row['Accuracy']:.2f}%"
    ax.text(i, row['Accuracy'] + 0.5, accuracy_value, ha='center', fontsize=10)
        
plt.tight_layout()

# Save the plot
plt.savefig('model_accuracy_comparison_no_winner_custom_colors.png')

print("model_accuracy_comparison_no_winner_custom_colors.png has been generated.")
```

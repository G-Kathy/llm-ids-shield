```python
import pandas as pd

df = pd.read_csv("hf://datasets/KathiS/new_preprocessed_dataset/new_cleaned.csv")

print(df)
```

```python
# -------------------------------
# Final 25 Most Important Features
# -------------------------------
important_features = [
    "Flow_Duration",
    "Dst_Port",
    "Flow_IAT_Min",
    "Flow_IAT_Mean",
    "Flow_IAT_Max",
    "Flow_Pkts/s",
    "Flow_Byts/s",
    "Pkt_Len_Max",
    "Pkt_Len_Mean",
    "Pkt_Size_Avg",
    "Pkt_Len_Min",
    "Pkt_Len_Std",          # or Pkt_Len_Var (choose 1)
    "Idle_Min",
    "Idle_Mean",
    "Idle_Max",
    "Fwd_Pkts/s",
    "Bwd_Pkts/s",
    "Init_Bwd_Win_Byts",
    "ACK_Flag_Cnt",
    "SYN_Flag_Cnt",
    "Bwd_Header_Len",
    "Fwd_Pkt_Len_Max",
    "Fwd_Pkt_Len_Min",
    "Bwd_Pkt_Len_Max",
    "Bwd_Pkt_Len_Mean"
]
```

```python
# Keep only these 25 features (and label column if present)
label_column = "Label"   # <-- change if your label name is different

if label_column in df.columns:
    final_columns = important_features + [label_column]
else:
    final_columns = important_features
```

```python
# Reduce dataset
df_selected = df[final_columns]
```

```python
# Save the selected feature dataset
df_selected.to_csv("Final_Preprocessed_File.csv", index=False)
```

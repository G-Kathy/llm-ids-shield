```python
import numpy as np
import matplotlib.pyplot as plt

# Hard-coded metrics (your provided values)
models = [
    "RandomForest",
    "XGBoost",
    "LightGBM (tuned)",
    "SVM"
]

metrics = ["Accuracy", "Precision", "Recall", "F1", "ROC AUC"]

values = np.array([
    [99.86, 99.85, 99.86, 99.85, 98.69],   # RandomForest
    [99.87, 99.86, 99.87, 99.86, 99.89],   # XGBoost
    [99.74, 99.74,  99.73, 99.73, 92.79],    # LightGBM tuned
    [86.33, 81.01, 86.33, 82.05, 77.79]    # SVM
])

# Plot settings
plt.figure(figsize=(14, 7))
bar_width = 0.16
x = np.arange(len(metrics))

# Updated colors based on previous instruction (assuming user wants 'Spectral' if they asked for 'more good' visuals)
# If you prefer the original colors, let me know!
import seaborn as sns
colors = [
    "#4C4B9B",   # Purple-blue
    "#356B8C",   # Steel blue
   
    "#4CAF75",   # Soft green
    "#B3CF58"    # Lime-yellow
]

# Plot each model
for i in range(len(models)):
    bars = plt.bar(x + i * bar_width, values[i], width=bar_width,
            label=models[i], color=colors[i])
    # Add exact percentage values on top of each bar
    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, yval + 1, round(yval, 2), ha='center', va='bottom', fontsize=8)

# Formatting
plt.xlabel("Performance Metrics", fontsize=14)
plt.ylabel("Score (%)", fontsize=14)
plt.title("Model Performance Comparison Across Metrics", fontsize=16, fontweight="bold")
plt.xticks(x + bar_width * 2, metrics, fontsize=12)
plt.ylim(0, 110)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.legend(title="Models", fontsize=12, bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0.) # Adjusted legend placement

plt.tight_layout(rect=[0, 0, 0.95, 1]) # Adjust layout to make space for the legend outside
plt.show()
```
